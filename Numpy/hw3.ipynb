```python
dept1 = np.array([100, 200, 300])
dept2 = np.array([400, 500, 600])
print(np.hstack((dept1,dept2)))
print(np.vstack((dept1,dept2)))
```

```python
a = np.array([[1, 2]])
b = np.array([[3, 4]])
print("Row:", np.concatenate((a,b),axis=0))
print("Column:", np.concatenate((a,b),axis=1))
```
