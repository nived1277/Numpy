```python
import numpy as np
arr1 = np.random.rand(10)
print(arr1)
mask1 = arr1 > 0.5
mask2 = arr1 <= 0.5
print("Present values:", arr1[mask1])
```

```python
arr2 = np.random.randint(0,100,size=15)
print("All scores:", arr2)
mas1 = arr2[arr2 > 60]
mas2 = arr2[arr2 % 5 == 0]
print("Score greater than 60:", mas1)
print("Score divisible by 5:", mas2)
```

```python
arr3 = np.array([100, 200, 300, 400, 500])
indices = [0,2,4]
print(arr3[indices])
```

```python
arr4 = np.random.randint(1,50,(4,4))
print(arr4)
print("Entire 2nd row:", arr4[1,:])
print("Entire 3rd column:", arr4[:,2])
```
