```python
import numpy as np
sales = np.array([250.75, 300.40, 275.60, 400.90, 150.30, 500.50, 350.80])
print("Total Week Sales:", np.sum(sales))
print("Mean Week Sales:", np.mean(sales))
print("Mediam Week Sales:", np.median(sales))
print("Std Dev Sales:", np.std(sales))
print("Max & Min Sales:", np.max(sales), "and", np.min(sales))

#Rounding & Mathematical Operations
print("\nRounding & Mathematical Operations")
print("Floor:", np.floor(sales))
print("Ceil:", np.ceil(sales))
print("Round:", np.round(sales))
expo = np.array([1, 2, 3])
print("Exponential Growth:", np.exp(expo))

#Cumulative Sales
print("\nCumulative Sales")
print("Cumilative Sales:", np.cumsum(sales))

#Matrix Operations
print("\nMatrix Operation")
A = np.array([[2, 1], [3, 4]])
B = np.array([[1, 2], [5, 6]])
print("Multiply:", A @ B)
print("Determinant:", np.linalg.det(A))
print("Inverse:", np.linalg.inv(A))
values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
C = np.array([[2,3],[1,4]])
D = np.array([5,11])
print("Leniar EQ Sol:", np.linalg.solve(C, D))
```
