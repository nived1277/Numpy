```python
import numpy as np
import pandas as pd
marks = np.array([45, 55, 60, 70, 85])
print("Added 5 Marks:", marks + 5)
normalized = ((marks - np.min(marks)) / (np.max(marks) - np.min(marks)))
print("Normalized Marks:", normalized) 
np.savetxt('marks.csv', marks, delimiter=',')
load = np.genfromtxt('marks.csv', delimiter=',')
print(load)
print(pd.DataFrame(marks, columns=["Normalized Marks"]))
```
