```python
import numpy as np
marks = np.array([35, 55, 75, 40, 90])
arr3 = np.where(marks >= 40, 'pass', 'fail')
print("Pass or Fail:", arr3)
```

```python
condition = [marks < 40, marks < 60, marks < 80]
choice = ['Fail', 'C', 'B']
print("Grade:", np.select(condition, choice, default='A'))
```

```python
marks2 = np.array([50, 60, 50, 70, 60, 80])
print("Unique Marks:", np.unique(marks2))
print("Any Failed:", np.any(marks2 < 40))
print("Check all Passed:", np.all(marks2 >= 40))
```
