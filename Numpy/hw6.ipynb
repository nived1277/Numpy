```python
import numpy as np
#salary data cleaning
salary = np.array([35000, 42000, np.nan, 50000, np.inf, 38000, np.nan])
print("Nan value:", np.isnan(salary))
print("Inf value:", np.isinf(salary))
salary[np.isinf(salary)] = 30000
salary[np.isnan(salary)] = 30000
print("cleaned array:", salary)

#Removing Invalid Data
salaries = np.array([35000, 42000, np.nan, 50000, np.inf, 38000, np.nan])
clean_salary = salaries[~np.isnan(salaries)]
clean_salary = clean_salary[~np.isinf(clean_salary)]
print("\nValid Salary:", clean_salary)

#copy() vs view()
arr = np.array([10, 20, 30, 40])
v = arr.view()
print("\nCopy:", arr.copy())
print("View:", v)
v[0] = 999
print(arr)
print(arr.view())

#Memory Concept (Thinking Task)

#Why modifying a view affects the original array ?
#-----> View shares ame memory as original array, So changes occured in view also affect in original array
#Why copy uses more memory ?
#-----> copy create complete seperate array in memory thats why copy uses more memory than views 
```
