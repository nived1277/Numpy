```python
import numpy as np
marks = np.array([78, 85, 62, 90, 55, 88])
print("Total Marks:", np.sum(marks))
print("Avg Marks:", np.mean(marks))
print("Median Marks:", np.median(marks))
print("Std Dev:", np.std(marks))
print("Min & Max Marks:", np.min(marks),"and", np.max(marks))
print("\nAfter adding 5 Grace Marks:")
print("Marks:", marks + 5)
print("\nAfter adding 10 prcntg bonus:")
print("Marks:", marks * 1.1)
```
