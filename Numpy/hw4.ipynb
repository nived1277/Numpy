```python
import numpy as np
scores =  np.array([25, 45, 65, 85, 95])
print("Condition:", np.where(scores < 40, 'Poor', 'Good'))
```

```python
condition = [scores < 40, scores < 70, scores < 90]
grade = ['Poor', 'Average', 'Good']
print("Grade:", np.select(condition, grade, default='Excellent'))
```

```python
arr1 = np.array([101, 102, 101, 103, 102, 104])
print("Unique:", np.unique(arr1))
```

```python
print("Check any high than 90:", np.any(scores > 90))
print("Check any less than 30:", np.all(scores < 30))
```
