```python
import numpy as np
marks_term1 = np.array([60, 70, 80, 90])
marks_term2 = np.array([65, 75, 85, 95])
print(marks_term1 + marks_term2)
print(marks_term2 - marks_term1)
print(marks_term1*1.05 , marks_term2*1.05)
```

```python
arr = np.array([marks_term1 + marks_term2])
print(arr > 150)
print((arr > 140) & (arr < 170))
```

```python
print("sum:", arr.sum())
print("mean:", arr.mean())
print("max:", arr.max())
print("min:", arr.min())
print("max:", arr.std())
```
